1) Define a function 'check_even_odd(number)' that takes a single input argument 'number' and checks whether it is even or odd using the bitwise operator. 

```python
def check_even_odd(number):
    if number & 1 == 0:
        return 'Even'
    else:
        return 'Odd'

print(check_even_odd(11))
```

2) Print numbers from 1 to 50 using 'for' loop.

   Rules:
* If number divisible by 2, print "Even"
* If number divisible by 5, Dont do anything
* If the number equals 23, stop the loop.

```python
def print_numbers(start, end):
    for i in range(start, end):
        if (i == 23):
            break
        elif (i % 5 == 0):
            continue
        elif (i % 2 == 0):
            print('even')
        else:
            print(i)

print_numbers(1, 51)
```

3) Write a program to reverse a number using 'while' loop

```python
def reverse_number (number):
    arr_number = list(str(number))
    i = len(arr_number)-1
    arr_reverse = []
    while(i >= 0):
        arr_reverse.append(arr_number[i])
        i-=1
    return int(str.join("", list(arr_reverse)))

reverse_number(989080948092384)
```

4) Define a function 'smiley()' that generates the coordinates of a smiley face.

Requirements:

The smiley face consists of:

* A unit circle centered at the origin representing the face.
* Two smaller circles of radius $0.2$ centered at $(-0.4,\,0.4)$ and $(0.4,\,0.4)$ representing the eyes.
* A circular arc representing the mouth, defined for $\theta \in \left[\frac{\pi}{6}, \frac{5\pi}{6}\right]$, with radius $0.6$ and shifted downward by $0.1$ units.

Combine all coordinates into two lists $x$ and $y$ and return them from the function.

Finally, call the function and use 'plt.plot' to display the smiley face.


```python
import numpy as np
import matplotlib.pyplot as plt

def circle(start, end, points, radius, cord_x, cord_y):
    theta = np.linspace(start, end, points)
    x = cord_x + radius * np.cos(theta)
    y = cord_y + radius * np.sin(theta)
    return x, y

def smiley():
    face_x, face_y = circle(0, 2*np.pi, 200, 1, 0, 0)
    left_eye_x, left_eye_y = circle(0, 2*np.pi, 200, 0.2, -0.4, 0.4)
    right_eye_x, right_eye_y = circle(0, 2*np.pi, 200, 0.2, 0.4, 0.4)
    mouth_x, mouth_y = circle(7*np.pi/6, 11*np.pi/6, 200, 0.6, 0, -0.1)

    x = np.concatenate([face_x, [np.nan], left_eye_x, [np.nan], right_eye_x, [np.nan], mouth_x])
    y = np.concatenate([face_y, [np.nan], left_eye_y, [np.nan], right_eye_y, [np.nan], mouth_y])
    return x, y

plt.scatter(0, 0)
plt.annotate('center (0,0)', (0.05,0))

x, y = smiley()
plt.plot(x,y)
plt.axis('equal')

plt.xlabel('x-axis')
plt.ylabel('y-axis')
```

5) Write a function 'ngon(n)' that generates the coordinates of a regular polygon with n sides.

Requirements:

The function must accept one parameter:

* n: an integer greater than 2.


The polygon should:

* Be centered at the origin $(0, 0)$
* Have a radius of 1 (unit circle)

The function must return two lists:

* $x$: $x$-coordinates of the vertices
* $y$: $y$-coordinates of the vertices

Finally, call the function and use 'plt.plot' to display the polygon.

```python
def ngon(n):
    if n <= 2:
        raise ValueError("n must be greater than 2")

    # Generate n equally spaced angles
    theta = np.linspace(0, 2*np.pi, n, endpoint=False)

    # Convert angles to coordinates (unit circle)
    x = np.cos(theta)
    y = np.sin(theta)

    # Close the polygon by repeating the first point
    x = np.append(x, x[0])
    y = np.append(y, y[0])
    print(y)
    return x, y

x, y = ngon(5)  # Example: pentagon

plt.plot(x, y);
plt.axis("equal")
```

6) Let $x$ be a list of points with the $x$-coordinates and $y$ be a lists of points with the $y$-coordinate. Suppose these coordinates are obtained from the function 'smiley()' or 'ngon()'.
Define a function 'translate($x, y, dx, dy$)' that translates each point by a displacement $dx$ in the $x$-direction and 
$dy$ in the $y$-direction. The function should return the new translated coordinate lists. Plot both the original and the translated coordinates of the smiley face or polygon on the same graph for comparison.

```python
def translate(x, y, dx, dy):
    x_d = x + dx
    y_d = y + dy
    return x_d, y_d

x, y = ngon(5)
plt.plot(x, y)

x_d, y_d = translate(x, y, 0.3, 0.2)
plt.plot(x_d, y_d)
plt.axis('equal')
```

7) Rotation of a any point $(x, y)$ in a two dimensional space about the $z$ axis can be performed by multiplying the point by a two-dimensional rotation matrix ($R$). If $\theta$ is the angle of rotation in radians, the matrix is given by, $ R = \begin{bmatrix} \cos \theta & -\sin \theta \\ \sin \theta & \cos \theta \end{bmatrix} $. Define a function 'rotate_point($x, y, \theta$)' that takes three input arguments $x$, $y$, and $ \theta $ and returns the rotated coordinates of the point ($x, y$) after rotation by the angle $\theta$ in radians. Then print both the original and the rotated point.

```python
def rotate_point(x, y, theta):
    x_rotate =  x * np.cos(theta) - y * np.sin(theta)
    y_rotate =  x * np.sin(theta) + y * np.cos(theta)
    return x_rotate, y_rotate

x, y = 0, 2
print('original points: x:{}, y:{}'.format(x, y))
plt.scatter(x, y)
plt.annotate('given point', (x, y))

x_rot, y_rot = rotate_point(x, y, np.pi)
print('rotated points: x:{}, y:{}'.format(x_rot, y_rot))
plt.scatter(x_rot, y_rot)
plt.annotate('rotated point', (x_rot, y_rot))
plt.xlabel('x')
plt.ylabel('y')

plt.xlim(-5, 5)
plt.ylim(-5, 5)
plt.grid(True)
```

8) Create a 'function rotate_list($x, y, \theta$)' to rotate a given set of coordinates in the two lists $x$, $y$ as in the previous problem and returns the rotated lists $x, y$ after rotation by the angle $\theta$ in radians. The input argument of function 'function rotate_list($x, y, \theta$)' are the lists $x$, $y$, which can be obtained from the function 'smiley()' or 'ngon()'. Then plot both the original and rotated smiley face or polygon on the same figure.

```python
def rotate_list(x, y, theta):
    x_rot = x * np.cos(theta) - y * np.sin(theta)
    y_rot = x * np.sin(theta) + y * np.cos(theta)
    return x_rot, y_rot

x, y = ngon(5)
plt.plot(x, y, label='original shape')

xr, yr = rotate_list(x, y, 2*np.pi/2)
plt.plot(xr, yr, label='rotated shape')

plt.axis('equal')
```

9) Consider the logistic map given by $y = r x (1-x)$ where $r$ is a constant between 0 and 4. The range of the $x$ values is between 0 and 1. Plot this function for 4 different $r$ values.

```python
def logistic_map(r):
    x = np.linspace(0, 1, 100)
    y = r*x*(1-x)
    return x, y

r_range = [1, 2, 3, 4]
for i in r_range:
    x, y = logistic_map(i)
    plt.plot(x, y)

plt.xlabel('x')
plt.ylabel('y')
```

)10) Read up about cobweb plots from here: https://en.wikipedia.org/wiki/Cobweb_plot and then implement a simple cobweb plot for the logistic map above for different values of $r$.

```python
def logistic_function(x, r):
    return r * x * (1 - x)

def cobweb_plot(r, x0, n_iter):
    # Create x values for smooth curve
    x = np.linspace(0, 1, 400)
    y = logistic_function(x, r)

    # Plot logistic curve
    plt.plot(x, y)

    # Plot diagonal y = x
    plt.plot(x, x)

    # Initial point
    x_n = x0

    for _ in range(n_iter):
        # Vertical line: (x_n, x_n) → (x_n, f(x_n))
        y_n = logistic_function(x_n, r)
        plt.plot([x_n, x_n], [x_n, y_n])

        # Horizontal line: (x_n, y_n) → (y_n, y_n)
        plt.plot([x_n, y_n], [y_n, y_n])

        # Update
        x_n = y_n

    plt.title(f"Cobweb Plot (r = {r})")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.xlim(0, 1)
    plt.ylim(0, 1)
    plt.grid(True)


r_values = [2.5, 3.2, 3.5, 3.9]

for r in r_values:
    plt.figure()
    cobweb_plot(r, x0=0.2, n_iter=50)

plt.show()
```
