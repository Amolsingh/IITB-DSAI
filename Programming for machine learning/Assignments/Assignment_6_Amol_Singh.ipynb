## Problem 2: Caesar Cipher Encoder

A *Caesar cipher* is one of the oldest encryption techniques. Every letter in a message is shifted by a fixed number of positions in the alphabet. For example, with a shift of 3:
- `'a'` → `'d'`
- `'z'` → `'c'` (wraps around)
- Non-letter characters (spaces, punctuation) stay the same.

This is a great exercise in combining string indexing, the modulo operator `%`, and building a new string character by character.

### Task
Complete the function `caesar_encode(text, shift)` that encodes a string using the Caesar cipher. The function must:
- Work for **both upper and lower case** letters.
- Leave non-letter characters unchanged.
- Wrap around the alphabet correctly using `%`.

### Expected output
```python
caesar_encode('Hello, World!', 3)  # 'Khoor, Zruog!'
caesar_encode('xyz', 3)            # 'abc'
caesar_encode('abc', 0)            # 'abc'
```

### Hint
- `ord('a')` gives the ASCII integer for `'a'` (97). `chr(97)` converts back.
- For a lowercase letter `c`, the shifted letter is:
  ```python
  chr((ord(c) - ord('a') + shift) % 26 + ord('a'))
  ```
- Use `c.isalpha()`, `c.islower()`, and `c.isupper()` to check character type.
- Build the result by concatenating characters into an output string.

## Problem 5: Mini Data Formatter

f-strings were introduced in lecture. There are some more additional feature f-string provides that are provided below.  
