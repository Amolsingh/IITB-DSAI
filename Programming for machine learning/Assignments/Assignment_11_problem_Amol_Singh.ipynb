### Task C — Create the standardization helper

Create `apply_standardization()`. This helper is for the manual preprocessing part. It should fit `StandardScaler` on training data only and transform test data using the same scaler.

#### Sample code:

```python
def apply_standardization(train_df, test_df, numeric_cols, target_col='price'):

    scaler = StandardScaler()

    cols_to_scale = [c for c in numeric_cols if c != target_col]

    train_df[cols_to_scale] = scaler.fit_transform(train_df[cols_to_scale])

    test_df[cols_to_scale] = scaler.transform(test_df[cols_to_scale])

    return train_df, test_df
```
