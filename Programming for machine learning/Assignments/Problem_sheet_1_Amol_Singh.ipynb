7. Create a list `temps` containing the following daily temperatures (in °C): `[22, 25, 19, 30, 28, 24, 18]`
   
    Using a loop compute the average temperature. \
    Using conditional statements, calculate the number of days with temperatures above average and print this count.

```python
temps = [22, 25, 19, 30, 28, 24, 18]

avg_temp = sum(temps) / len(temps)

print("Average temprature in (C): {}".format(avg_temp))

number_of_days_above_average = 0

for i in temps:
    if(i>avg_temp):
        number_of_days_above_average += 1

print("Number of days above average: {}".format(number_of_days_above_average))
```

    Create one more empty list `grade`:\
    Using a loop, assign the grades based on the following condition
    
    - "A" if mark ≥ 85
    - "B" if 70 ≤ mark < 85
    - "C" if 50 ≤ mark < 70
    - "F" if mark < 50
    
Using a loop, print each student’s name, mark, and grade.\
Count and print how many students passed (mark ≥ 50).

```python
students = ["Asha", "Ben", "Carlos", "Divya", "Evan"]
marks = [72, 85, 64, 90, 58]
grade = []

for i in marks:
    match i:
        case m if (m>=85):
            grade.append('A')
        case m if (m>=70 and m<85):
            grade.append('B')
        case m if (m>=50 and m<70):
            grade.append('C')
        case m if (m<50):
            grade.append('D')

def students_passed(marks):
    count = 0
    for i in marks:
        if(i >= 50):
            count += 1
    return count

for i in range(len(marks)):
    print('{s}, {m}, {g}'.format(s=students[i], m=marks[i], g=grade[i]))

print('The number of students passed: {}'.format(students_passed(marks)))
```

9. Create a list `prices` = `[1200, 750, 2999, 499, 850]`

    Write a function `apply_discount(price)` that:
    
    - applies a **20%** discount if **price** > 1000
    - applies a **10%** discount otherwise
    - returns the final price

Using a loop, apply the discount to each item and store the discounted prices in a new list `final_price`.\
Compute and print the total cost after discounts by summing over the `final_price` list.

```python
prices = [1200, 750, 2999, 499, 850]
print(prices)

def apply_discount(price):
    match price:
        case p if (p>1000):
            return (p - (p*20/100))
        case p if(p<1000):
            return (p - (p*10/100))

final_price = []
for price in prices:
    final_price.append(apply_discount(price))

print(final_price)

print("The final price is: {}".format(sum(final_price)))

```

10. Write a function `is_prime(n)` that:

    - returns `True` if **n** is a prime number
    - returns `False` otherwise
    
Create a list of numbers from 1 to 100.\
Using a loop and the function, create a list of all prime numbers.\
Count how many prime numbers are in the list.\
Print the largest prime number found.

```python
import math

def is_prime(n):
    if (n <= 1):
        return False
    if (n == 2):
        return True
    for i in range(2, int(math.sqrt(n)) + 1):
        if (n % i) == 0:
            return False
    return True

prime = []
for i in range(1, 100):
    if is_prime(i):
        prime.append(i)

print(prime)
print("Count of prime numbers: {}".format(len(prime)))
prime.sort()
prime[len(prime)-1]
```
